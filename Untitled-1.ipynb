1. Defining the Vessel

```python
class Vessel:
    def __init__(self, load, length, arrival_time, status):
        self.load = load
        self.length = length
        self.arrival_time = arrival_time  # relative offset; negative = already arrived
        self.status = status

    def __repr__(self):
        return (
            f"[load={self.load}, length={self.length}, "
            f"arrival={self.arrival_time}, status={self.status}]"
        )
```

2. Managing the Queue

```python
from collections import deque
import random

class VesselQueue:
    def __init__(self, capacity=10, arrival_scale=4.0):
        # We manage capacity manually to match the service logic.
        self.queue = deque()
        self.capacity = capacity
        self.current_time = 0
        self.arrival_scale = arrival_scale
        # Start slightly in the past so some vessels are already arrived
        self.last_eta = -random.randint(0, 5)

        # Pre-fill the queue with vessels to start
        for _ in range(capacity):
            self._enqueue_internal(self._generate_random_vessel())

        self._update_statuses()

    def _next_eta(self):
        interarrival = max(1, int(round(random.expovariate(1 / self.arrival_scale))))
        self.last_eta += interarrival
        return self.last_eta

    def _generate_random_vessel(self):
        """Helper to create a new vessel with random attributes."""
        load = random.randint(200, 2000)  # workload in TEU-like units
        length = random.randint(2, 10)    # length in quay blocks
        eta = self._next_eta()
        arrival_time = eta - self.current_time  # relative offset
        status = "on_the_way" if arrival_time > 0 else "waiting"
        return Vessel(
            load=load,
            length=length,
            arrival_time=arrival_time,
            status=status
        )

    def _enqueue_internal(self, vessel):
        if len(self.queue) >= self.capacity:
            print("Queue is full! Vessel rejected.")
            return False
        self.queue.append(vessel)
        return True

    def enqueue(self, vessel):
        """Public enqueue (kept for completeness)."""
        return self._enqueue_internal(vessel)

    def dequeue(self):
        if not self.queue:
            return None
        return self.queue.popleft()

    def _advance_time(self, steps=1):
        self.current_time += steps
        for vessel in self.queue:
            vessel.arrival_time -= steps
        self._update_statuses()

    def advance_time(self, steps=1):
        """Public time advance for demos."""
        self._advance_time(steps)

    def _update_statuses(self):
        for vessel in self.queue:
            if vessel.arrival_time <= 0:
                vessel.status = "waiting"
            else:
                vessel.status = "on_the_way"

    def mark_at_berth(self, index):
        if index < 0 or index >= len(self.queue):
            print("Invalid vessel index.")
            return
        vessel = self.queue[index]
        if vessel.arrival_time > 0:
            print("Selected vessel has not arrived yet.")
            return
        vessel.status = "at_berth"
        print(f"Updated vessel to at_berth: {vessel}")

    def service_next_vessel(self):
        if not self.queue:
            print("Queue is empty!")
            return

        # FIFO: service the vessel at the front
        candidate = self.queue[0]
        if candidate.arrival_time > 0:
            print("Front vessel has not arrived yet.")
            self._advance_time()
            return

        finished_vessel = self.dequeue()
        finished_vessel.status = "at_berth"
        print(f"Serviced (FIFO): {finished_vessel}")

        # New vessel enters at the back
        new_vessel = self._generate_random_vessel()
        self._enqueue_internal(new_vessel)
        print(f"New vessel entered: {new_vessel}")

        self._advance_time()

    def service_vessel_at(self, index):
        if not self.queue:
            print("Queue is empty!")
            return
        if index < 0 or index >= len(self.queue):
            print("Invalid vessel index.")
            return

        candidate = self.queue[index]
        if candidate.arrival_time > 0:
            print("Selected vessel has not arrived yet.")
            self._advance_time()
            return

        finished_vessel = self.queue[index]
        del self.queue[index]
        finished_vessel.status = "at_berth"
        print(f"Serviced (agent pick): {finished_vessel}")

        # New vessel enters at the back
        new_vessel = self._generate_random_vessel()
        self._enqueue_internal(new_vessel)
        print(f"New vessel entered: {new_vessel}")

        self._advance_time()

    def show_queue(self):
        print("--- Current Vessel Queue ---")
        for i, vessel in enumerate(self.queue):
            print(f"{i+1}. {vessel}")
        print(f"Time tick: {self.current_time}")
        print("----------------------------")
```

3. How it Works in Practice

```python
# Initialize the queue with 10 vessels
port_queue = VesselQueue(capacity=10)
port_queue.show_queue()

# Service a few vessels (FIFO)
for _ in range(2):
    port_queue.service_next_vessel()

# Agent picks a vessel by index (0-based)
port_queue.service_vessel_at(4)

# See the updated state
port_queue.show_queue()
```

4. Demo: Vessel Disappears + One Updated


```python
# Ensure some vessels have arrived
target_steps = 3
port_queue.advance_time(target_steps)

# Update one vessel's status to at_berth
port_queue.mark_at_berth(0)

# Remove a vessel via servicing (disappears) and append a new one
port_queue.service_vessel_at(1)

# Show queue order and arrival offsets (should be non-decreasing)
port_queue.show_queue()
print("Arrival offsets in queue order:", [v.arrival_time for v in port_queue.queue])
```
