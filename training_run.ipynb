# Training Run Notebook

Notebook-based training entrypoint for the Antwerp port RL system.

```python
from training.config import load_config
from training.model import train_model
from training.eval import evaluate_model

cfg = load_config("training_config.json")
seed = int(cfg["seed"])

# Short run for notebook iteration
model, _, _ = train_model(cfg, seed_override=seed, timesteps_override=5000)
model.save(cfg["model_save_path"])
print("Saved model to:", cfg["model_save_path"])
```

```python
env_cfg = dict(cfg.get("env", {}))
mean_reward, mean_length = evaluate_model(model, seed + 1000, env_cfg, episodes=5)
print(f"Mean reward: {mean_reward:.3f}")
print(f"Mean episode length: {mean_length:.2f}")
```
